```python

# ============================================================
# АНАЛИЗ ЛУЧШЕГО ВРЕМЕНИ ДЛЯ ПУБЛИКАЦИЙ TELEGRAM-КАНАЛА
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

from google.colab import files

# ------------------------------------------------------------
# 1. ЗАГРУЗКА ФАЙЛА
# ------------------------------------------------------------

file_path = '/Besfiltrov .csv'

df = pd.read_csv(
    file_path,
    sep=';',
    encoding='utf-8-sig'
)

print("Файл успешно загружен!")
print("Количество записей:", len(df))
print("Столбцы:", df.columns.tolist())


# ------------------------------------------------------------
# 2. ПРЕДОБРАБОТКА ДАННЫХ
# ------------------------------------------------------------

data = df.copy()

# Преобразуем дату публикации
data['date'] = pd.to_datetime(
    data['date'],
    utc=True,
    errors='coerce'
)

data = data.dropna(subset=['date']).copy()

# Переводим время в московский часовой пояс
data['date'] = data['date'].dt.tz_convert('Europe/Moscow')

# Преобразуем числовые показатели
numeric_columns = [
    'views',
    'comments',
    'reactions_total',
    'forwards'
]

for col in numeric_columns:
    data[col] = pd.to_numeric(
        data[col], errors='coerce'
    ).fillna(0)

# Извлекаем час и день недели
data['hour'] = data['date'].dt.hour
data['weekday'] = data['date'].dt.dayofweek

days = {
    0: 'Понедельник',
    1: 'Вторник',
    2: 'Среда',
    3: 'Четверг',
    4: 'Пятница',
    5: 'Суббота',
    6: 'Воскресенье'
}

data['day_name'] = data['weekday'].map(days)


# ------------------------------------------------------------
# 3. ОБЪЕДИНЕНИЕ ЗАПИСЕЙ ОДНОЙ ПУБЛИКАЦИИ
# ------------------------------------------------------------

posts = (
    data.groupby('date', as_index=False)
    .agg({
        'views': 'max',
        'comments': 'max',
        'reactions_total': 'max',
        'forwards': 'max',
        'hour': 'first',
        'weekday': 'first',
        'day_name': 'first'
    })
)

print("\nКоличество публикаций после объединения:", len(posts))


# ------------------------------------------------------------
# 4. ВЫБОР ПЕРИОДА АНАЛИЗА
# ------------------------------------------------------------

# Анализируем последние 12 месяцев
analysis_months = 12

end_date = posts['date'].max()
start_date = end_date - pd.DateOffset(months=analysis_months)

analysis = posts[
    (posts['date'] >= start_date) &
    (posts['date'] <= end_date)
].copy()

print("\nПЕРИОД АНАЛИЗА")
print("Начало:", start_date)
print("Окончание:", end_date)
print("Количество публикаций:", len(analysis))


# ------------------------------------------------------------
# 5. АНАЛИЗ ПО ДНЯМ НЕДЕЛИ
# ------------------------------------------------------------

weekday_stats = (
    analysis.groupby(['weekday', 'day_name'])
    .agg(
        posts_count=('views', 'count'),
        avg_views=('views', 'mean'),
        median_views=('views', 'median'),
        avg_reactions=('reactions_total', 'mean'),
        avg_comments=('comments', 'mean'),
        avg_forwards=('forwards', 'mean')
    )
    .reset_index()
    .sort_values('weekday')
)

print("\nАКТИВНОСТЬ ПО ДНЯМ НЕДЕЛИ")
display(weekday_stats.round(2))

# График по дням недели
plt.figure(figsize=(12, 6))

sns.barplot(
    data=weekday_stats,
    x='day_name',
    y='avg_views'
)

plt.title('Среднее количество просмотров по дням недели')
plt.xlabel('День недели')
plt.ylabel('Среднее количество просмотров')
plt.xticks(rotation=30)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 6. АНАЛИЗ ПО ЧАСАМ
# ------------------------------------------------------------

hour_stats = (
    analysis.groupby('hour')
    .agg(
        posts_count=('views', 'count'),
        avg_views=('views', 'mean'),
        median_views=('views', 'median'),
        avg_reactions=('reactions_total', 'mean'),
        avg_comments=('comments', 'mean'),
        avg_forwards=('forwards', 'mean')
    )
    .reset_index()
)

print("\nАКТИВНОСТЬ ПО ЧАСАМ")
display(hour_stats.round(2))

# График по часам
plt.figure(figsize=(14, 6))

sns.barplot(
    data=hour_stats,
    x='hour',
    y='avg_views'
)

plt.title('Среднее количество просмотров по часам')
plt.xlabel('Час публикации (МСК)')
plt.ylabel('Среднее количество просмотров')
plt.xticks(range(24))
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 7. ТЕПЛОВАЯ КАРТА: ДЕНЬ НЕДЕЛИ И ЧАС
# ------------------------------------------------------------

heatmap_data = analysis.pivot_table(
    index='day_name',
    columns='hour',
    values='views',
    aggfunc='mean'
)

heatmap_data = heatmap_data.reindex(list(days.values()))

plt.figure(figsize=(18, 7))

sns.heatmap(
    heatmap_data,
    cmap='YlOrRd',
    annot=True,
    fmt='.0f',
    linewidths=0.5,
    cbar_kws={'label': 'Среднее количество просмотров'}
)

plt.title('Популярность публикаций: день недели и час')
plt.xlabel('Час публикации (МСК)')
plt.ylabel('День недели')
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 8. АНАЛИЗ РЕАКЦИЙ, КОММЕНТАРИЕВ И ПЕРЕСЫЛОК
# ------------------------------------------------------------

fig, axes = plt.subplots(1, 3, figsize=(19, 5))

metrics = [
    ('avg_reactions', 'Среднее количество реакций'),
    ('avg_comments', 'Среднее количество комментариев'),
    ('avg_forwards', 'Среднее количество пересылок')
]

for ax, (metric, title) in zip(axes, metrics):

    sns.barplot(
        data=hour_stats,
        x='hour',
        y=metric,
        ax=ax
    )

    ax.set_title(title)
    ax.set_xlabel('Час публикации')
    ax.set_ylabel('Количество')
    ax.set_xticks(range(24))
    ax.tick_params(axis='x', labelrotation=45)
    ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 9. ПОИСК ЛУЧШЕГО ВРЕМЕНИ ПУБЛИКАЦИИ
# ------------------------------------------------------------

best_times = (
    analysis.groupby(['weekday', 'day_name', 'hour'])
    .agg(
        posts_count=('views', 'count'),
        avg_views=('views', 'mean'),
        median_views=('views', 'median'),
        avg_reactions=('reactions_total', 'mean'),
        avg_comments=('comments', 'mean'),
        avg_forwards=('forwards', 'mean')
    )
    .reset_index()
)

# Учитываем только интервалы с минимум 5 публикациями
reliable_times = best_times[
    best_times['posts_count'] >= 5
].copy()

best_times_sorted = reliable_times.sort_values(
    'avg_views',
    ascending=False
)

print("\nТОП-10 ЛУЧШИХ ВРЕМЕННЫХ ИНТЕРВАЛОВ")
display(best_times_sorted.head(10).round(2))

# Выводим лучший временной интервал
if len(best_times_sorted) > 0:

    best = best_times_sorted.iloc[0]

    print("\nРЕКОМЕНДУЕМОЕ ВРЕМЯ ПУБЛИКАЦИИ")
    print("--------------------------------")
    print("День:", best['day_name'])
    print("Время:", str(int(best['hour'])) + ":00")
    print("Средние просмотры:", round(best['avg_views']))
    print("Количество публикаций:", int(best['posts_count']))

else:
    print("\nНедостаточно публикаций для определения")
    print("надежного сочетания дня и часа.")


# ------------------------------------------------------------
# 10. СОХРАНЕНИЕ РЕЗУЛЬТАТОВ В EXCEL
# ------------------------------------------------------------

output_file = 'Результаты_анализа_БЕЗФИЛЬТРОВ.xlsx'

with pd.ExcelWriter(output_file) as writer:

    weekday_stats.to_excel(
        writer,
        sheet_name='Дни недели',
        index=False
    )

    hour_stats.to_excel(
        writer,
        sheet_name='Часы',
        index=False
    )

    best_times_sorted.to_excel(
        writer,
        sheet_name='Лучшее время',
        index=False
    )

print("\nАНАЛИЗ ЗАВЕРШЁН!")
print("Результаты сохранены в Excel.")

files.download(output_file)
```
